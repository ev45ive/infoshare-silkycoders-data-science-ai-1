# Sprzedaż według kanału
Źródło: dzienny wyciąg pozycji sprzedaży.

```python
import csv
from collections import defaultdict
from pathlib import Path

source = Path('../datasets/sales.csv')
result = defaultdict(int)
with source.open(encoding='utf-8', newline='') as f:
    for row in csv.DictReader(f):
        result[row['channel']] += int(row['quantity'])
for channel, units in sorted(result.items()):
    print(channel, units)
```
